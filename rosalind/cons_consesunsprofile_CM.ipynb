# Consensus and Profile 

https://rosalind.info/problems/cons/ <br/>

In the HAMM exercise (counting point mutations), we calculated the minimum number of symbols mismatches between the string `s` and `t`. If instead we have several homologous strands to analyse simoultaneously, then the natural provlem is to find an avarage-case strand to represent the most likely common ancestor of the given strands. 

- Given: a list of DNA strings which we then create a `matrix` from, with intersection value of row i and j &rarr; Aij.
  - translate into the number of values for each A C G T in each column &rarr; Pij.
  - create a string of lenght n formed from the collection by taking the most common symbol at each position &rarr; `consensus string`
- Result: a consensus string and a profile matrix for the collection.

```python
from util import read_file_into_lines, read_fasta, text_to_fasta
import numpy as np
```

```python
fasta = read_fasta('cons_trial.txt')
```

```python
def create_profile(fasta):
    matrix = []
    for v in fasta.values():
        curr = list(v)
        matrix.append(curr)
    matrix = np.asarray(matrix)
    
    profile = {}
    count_A = ''
    count_C = ''
    count_G = ''
    count_T = ''
    for i in range(len(matrix[0])):
        count_A += str(np.count_nonzero(matrix[:,i] == 'A')) + ' '
        count_C += str(np.count_nonzero(matrix[:,i] == 'C')) + ' '
        count_G += str(np.count_nonzero(matrix[:,i] == 'G')) + ' '
        count_T += str(np.count_nonzero(matrix[:,i] == 'T')) + ' '
    profile['A:'] = count_A
    profile['C:'] = count_C
    profile['G:'] = count_G
    profile['T:'] = count_T
    return(profile)
```

```python
def create_consesus(profile):
    prelim = []
    for k, v in profile.items():
        curr = k[0] + v.replace(' ', '')
        prelim.append(list(curr))
    prelim = np.asarray(prelim)
    
    positions = ''
    for i in range(1, len(prelim[0])):
        curr_max = str(max(prelim[:, i]))
        curr_string = "".join(str(x) for x in prelim[:, i])
        positions += str(curr_string.find(curr_max))
    consensus = positions.replace('0', 'A').replace('1', 'C').replace('2', 'G').replace('3', 'T')
    return(consensus)
```

```python
profile = create_profile(fasta)
consensus = create_consesus(profile)

print(consensus) 
for k, v in profile.items():
    print(k, v)
```

# Solve exercise

```python
fasta = read_fasta('rosalind_cons.txt')
profile = create_profile(fasta)
consensus = create_consesus(profile)

print(consensus) 
for k, v in profile.items():
    print(k, v)
```
